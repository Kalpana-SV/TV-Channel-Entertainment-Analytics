```python
for col in df.columns:
    print("\n", col)
    print(df[col].astype(str).str.strip().str.lower().isin(
        ["na", "n/a", "unknown", ""]
    ).sum())
```

```python
for col in df.columns:
    converted = pd.to_numeric(df[col], errors="coerce")
    if converted.notna().sum() > 0:
        print(col, "->", df[col].dtype)
```

```python
df["Channel_Name"] = df["Channel_Name"].astype(str).str.strip().str.title()

print(df["Channel_Name"].unique())
```

```python
print(df.isnull().sum())
num_cols = df.select_dtypes(include="number").columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
cat_cols = df.select_dtypes(include="object").columns

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df = df.drop_duplicates()

print("Duplicates:", df.duplicated().sum())
```

```python
df.loc[
    (df["Avg_Rating"] < 0) | (df["Avg_Rating"] > 5),
    "Avg_Rating"
] = df["Avg_Rating"].median()
```

```python
df.loc[
    df["Subscription_Fee"] < 0,
    "Subscription_Fee"
] = df["Subscription_Fee"].median()
```

```python
df.loc[
    (df["Years_Active"] < 0) | (df["Years_Active"] > 100),
    "Years_Active"
] = df["Years_Active"].median()
```

```python
numeric_columns = [
    "Avg_Viewers",
    "Avg_Rating",
    "Total_Likes",
    "YouTube_Views",
    "Social_Media_Followers",
    "Subscription_Fee",
    "Years_Active",
    "Monthly_Reach"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df = df.replace(
    ["NA", "N/A", "na", "n/a", "unknown", "Unknown", ""],
    pd.NA
)

print(df.isnull().sum())
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate records:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
df.to_csv("cleaned_tv_channel_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
